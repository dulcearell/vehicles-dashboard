# Análisis exploratorio de anuncios de vehículos

Este notebook realiza una exploración básica de `vehicles_us.csv` y crea las visualizaciones que servirán como base para la aplicación web.

```python
import pandas as pd
import plotly.express as px
```

```python
car_data = pd.read_csv('../vehicles_us.csv')
car_data.head()
```

## Estructura y calidad de los datos

```python
car_data.info()
car_data.describe(include='all').T
```

```python
missing_values = car_data.isna().sum().sort_values(ascending=False)
missing_values[missing_values > 0]
```

El conjunto contiene valores ausentes, especialmente en `is_4wd`, `paint_color`, `odometer`, `cylinders` y `model_year`. Para esta primera versión no se eliminan ni imputan globalmente; cada visualización omite únicamente los datos faltantes que necesita.

## Distribución del kilometraje

```python
fig_hist = px.histogram(
    car_data, x='odometer', nbins=50,
    title='Distribución del kilometraje',
    labels={'odometer': 'Kilometraje'}
)
fig_hist.update_layout(yaxis_title='Número de anuncios')
fig_hist.show()
```

## Relación entre precio y kilometraje

```python
scatter_data = car_data.dropna(subset=['odometer', 'price', 'condition'])
fig_scatter = px.scatter(
    scatter_data, x='odometer', y='price', color='condition', opacity=0.5,
    title='Precio y kilometraje de los vehículos',
    labels={'odometer': 'Kilometraje', 'price': 'Precio (USD)', 'condition': 'Condición'}
)
fig_scatter.show()
```

## Conclusiones preliminares

- El conjunto contiene más de 50 mil anuncios de vehículos.
- El kilometraje presenta una distribución amplia y algunos valores extremos.
- El gráfico de dispersión permite observar cómo varía el precio conforme aumenta el kilometraje y comparar las distintas condiciones de los vehículos.
- Antes de realizar modelos o conclusiones estadísticas más profundas sería necesario estudiar y tratar los valores ausentes y atípicos.
